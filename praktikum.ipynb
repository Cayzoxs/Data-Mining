## 1. Konsep Dasar Data Mining

Data Mining adalah proses menemukan pola, hubungan, atau informasi yang berguna dari sekumpulan data. Hasilnya dapat membantu seseorang memahami kondisi yang terjadi dan membuat keputusan yang lebih baik.

Contoh penerapan:
- **E-commerce:** merekomendasikan produk berdasarkan riwayat pembelian.
- **Perbankan:** membantu mendeteksi transaksi yang tidak biasa.
- **Pendidikan:** menganalisis faktor yang berhubungan dengan hasil belajar.
- **Kesehatan:** mempelajari pola data untuk mendukung analisis tenaga kesehatan.

Data Mining bukan sekadar membuat grafik atau mengumpulkan data. Intinya adalah mencari informasi yang bermakna dari data, lalu menafsirkan hasilnya secara hati-hati.

## 2. Pengenalan CRISP-DM

CRISP-DM adalah kerangka kerja yang sering digunakan untuk mengatur proses proyek Data Mining. Secara umum, tahapannya adalah:

1. **Business Understanding** — memahami masalah dan tujuan.
2. **Data Understanding** — mengumpulkan dan memahami data.
3. **Data Preparation** — membersihkan dan menyiapkan data.
4. **Modeling** — menerapkan metode atau algoritma yang sesuai.
5. **Evaluation** — menilai apakah hasil menjawab tujuan awal.
6. **Deployment** — menyampaikan atau menerapkan hasil agar bisa digunakan.

Tahapan ini tidak selalu berjalan lurus satu kali. Ketika hasil evaluasi belum memadai, proses dapat kembali ke tahap sebelumnya.

## 3. Mengenal Data dengan Python

Kita akan menggunakan data transaksi toko fiktif berukuran kecil. Pada tahap pengantar ini, kita hanya mengamati struktur dan pola sederhana; belum membangun model prediksi.

```python
import pandas as pd

data = {
    "transaksi": ["T001", "T002", "T003", "T004", "T005", "T006", "T007", "T008"],
    "kategori_produk": ["Makanan", "Minuman", "Makanan", "Perawatan", "Minuman", "Makanan", "Perawatan", "Minuman"],
    "jumlah_item": [3, 2, 5, 1, 4, 2, 1, 3],
    "total_belanja_ribu": [75, 30, 120, 90, 55, 60, 110, 45]
}

df = pd.DataFrame(data)
df
```

### Apa yang dilakukan kode di atas?
- `import pandas as pd` memanggil pustaka pandas untuk mengolah data berbentuk tabel.
- `data` menyimpan contoh data dalam bentuk kolom dan nilai.
- `pd.DataFrame(data)` mengubah data tersebut menjadi tabel.
- `df` adalah variabel yang menyimpan tabel tersebut.

Angka pada kolom `total_belanja_ribu` menggunakan satuan ribuan rupiah. Contohnya, nilai `75` berarti Rp75.000.
